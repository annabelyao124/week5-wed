### Inputs for the problem

````python
documents = [
    "Max likes Sam.",    # D1
    "Sam likes Max.",    # D2
    "Max and Sam.",      # D3
    "Winter likes fall." # D4
]

query = "Sam likes winter"
````

### Step 1: Create vocabulary of all terms

First I will concatenate all strings into one.

````python
allText = " ".join(documents + [query])
print(allText)
````

I don't want to have punctuation. There are several ways to do that. Let's see some:

#### Method 1: Iterate over the string, leave out punctuations

````python
import string
print(string.punctuation)
````

````python
allText_noPunct = "".join([ch for ch in allText if not ch in string.punctuation])
print(allText_noPunct)
````

### Method 2: Use built-in methods

Pythyn's built-in `str` class has many useful methods for translating a set of characters to another one. Let's see the method in action:

````python
allText.translate(str.maketrans("", "", string.punctuation))
````

What's going on?

````python
print(str.maketrans("", "", string.punctuation))
````

We can see that all ASCII Characters from 0-32 and from 65 to 90 are missing, because those characters will be deleted.

Here is what is happening:

```
str.maketrans(from_str, to_str, delete_chars)
#              ^          ^        ^
#           1st arg    2nd arg  3rd arg
```

* 1st argument (""): The characters you want to replace. (Empty string = replace nothing)

* 2nd argument (""): The characters to replace them with. (Empty string = no replacement mapping)

* 3rd argument (string.punctuation): The characters you want to delete entirely.

### Our vocabulary

Here is a Python one-liner that takes our variable `allText_noPunct` and find the unique words in it.

````python
words = set(allText_noPunct.lower().split())
words
````

````python
voc = sorted(words)
voc
````

### Step 2: Represent each document and the query in the vector space created by these terms

There are total of 6 words, and for each document, we represent the total number that each word appears.

````python
# lowercase, remove punctuation, and split into words
text = documents[0]                                         
text = text.lower()                          
text = text.translate(str.maketrans("", "", string.punctuation))
d1_words = text.split()
print(d1_words)
````

````python
# Go through each word and count its occurences in D1
print("Vocabulary:", voc)
print()

d1_vector = []
for word in voc:
    count = d1_words.count(word)
    print(word, "appears", count, "time(s) in D1")
    d1_vector.append(count)

print()
print("D1 vector:", d1_vector)
````

````python
# Do the same for every document and the query. First clean up the word, then count the occurences.
# Output the document-term matrix and the query vector 
doc_names = ["D1", "D2", "D3", "D4"]
doc_term_matrix = []         

for doc in documents:
    text = doc.lower()
    text = text.translate(str.maketrans("", "", string.punctuation))
    words_in_doc = text.split()

    vector = []
    for word in voc:
        vector.append(words_in_doc.count(word))

    doc_term_matrix.append(vector)

text = query.lower()
text = text.translate(str.maketrans("", "", string.punctuation))
words_in_query = text.split()

query_vector = []
for word in voc:
    query_vector.append(words_in_query.count(word))

# show the matrix with its labels
for name, vector in zip(doc_names, doc_term_matrix):
    print(name, "    ", vector)
print("Query ", " ", query_vector)
````

### Step 3: Calculate the cosine similarity between the query and each document

````python
# distance for A and B
sum_of_squares_A = 0
for a in A:
    sum_of_squares_A = sum_of_squares_A + a * a
length_A = sum_of_squares_A ** 0.5

sum_of_squares_B = 0
for b in B:
    sum_of_squares_B = sum_of_squares_B + b * b
length_B = sum_of_squares_B ** 0.5

print("Query: sum of squares =", sum_of_squares_A, " length =", length_A)
print("D1   : sum of squares =", sum_of_squares_B, " length =", length_B)
````
